# 01 – Análisis Exploratorio de Datos (EDA)

Este notebook realiza la exploración inicial del conjunto de datos:
- Carga y resumen del dataset
- Distribución de variables
- Detección de valores faltantes
- Análisis de correlaciones
- Visualizaciones exploratorias

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sys
import os

# Agregar la raíz del proyecto al path
sys.path.insert(0, os.path.abspath('..'))
from src.data_loader import cargar_datos, resumen_datos

sns.set_theme(style='whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)
```

## 1. Carga del conjunto de datos

> **Instrucción:** Coloca tu archivo de datos en `data/raw/` y actualiza la ruta a continuación.

```python
# --- Ejemplo con el dataset Iris incluido en scikit-learn ---
from sklearn.datasets import load_iris

iris = load_iris(as_frame=True)
df = iris.frame
df['target_name'] = df['target'].map(dict(enumerate(iris.target_names)))

# Para usar tu propio dataset, descomenta la línea siguiente:
# df = cargar_datos('../data/raw/tu_archivo.csv')

df.head()
```

## 2. Resumen general

```python
resumen_datos(df)
```

## 3. Distribución de la variable objetivo

```python
columna_objetivo = 'target'  # Cambia al nombre de tu columna objetivo

plt.figure()
df[columna_objetivo].value_counts().plot(kind='bar', color='steelblue', edgecolor='black')
plt.title('Distribución de la variable objetivo')
plt.xlabel('Clase')
plt.ylabel('Frecuencia')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('../reports/figures/distribucion_objetivo.png', dpi=150)
plt.show()
```

## 4. Distribución de variables numéricas

```python
numericas = df.select_dtypes(include=[np.number]).columns.tolist()

df[numericas].hist(bins=20, figsize=(14, 8), edgecolor='black', color='steelblue')
plt.suptitle('Histogramas de variables numéricas', fontsize=14)
plt.tight_layout()
plt.savefig('../reports/figures/histogramas_numericas.png', dpi=150)
plt.show()
```

## 5. Análisis de valores faltantes

```python
faltantes = df.isnull().sum().sort_values(ascending=False)
faltantes_pct = (faltantes / len(df) * 100).round(2)
resumen_faltantes = pd.DataFrame({'Valores Nulos': faltantes, '% Nulos': faltantes_pct})
print(resumen_faltantes[resumen_faltantes['Valores Nulos'] > 0])
```

## 6. Matriz de correlación

```python
plt.figure(figsize=(10, 8))
corr = df[numericas].corr()
mask = np.triu(np.ones_like(corr, dtype=bool))
sns.heatmap(corr, mask=mask, annot=True, fmt='.2f', cmap='coolwarm',
            square=True, linewidths=0.5)
plt.title('Matriz de Correlación')
plt.tight_layout()
plt.savefig('../reports/figures/matriz_correlacion.png', dpi=150)
plt.show()
```

## 7. Pairplot (relaciones entre variables)

```python
g = sns.pairplot(df, hue='target_name', diag_kind='kde', plot_kws={'alpha': 0.6})
g.fig.suptitle('Pairplot del conjunto de datos', y=1.02, fontsize=14)
plt.savefig('../reports/figures/pairplot.png', dpi=150)
plt.show()
```

## 8. Boxplots por clase

```python
cols_numericas = [c for c in numericas if c != columna_objetivo]
fig, axes = plt.subplots(1, len(cols_numericas), figsize=(14, 5))

for ax, col in zip(axes, cols_numericas):
    sns.boxplot(data=df, x='target_name', y=col, palette='Set2', ax=ax)
    ax.set_title(col)
    ax.set_xlabel('')

plt.suptitle('Distribución por clase', fontsize=14)
plt.tight_layout()
plt.savefig('../reports/figures/boxplots_por_clase.png', dpi=150)
plt.show()
```
