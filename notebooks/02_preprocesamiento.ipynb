# 02 – Preprocesamiento de Datos

Este notebook aplica las transformaciones necesarias para preparar el conjunto de datos antes del entrenamiento:
- Separación de características y variable objetivo
- División en conjuntos de entrenamiento y prueba
- Imputación de valores faltantes
- Codificación de variables categóricas
- Escalado de características
- Exportación del conjunto preprocesado

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sys
import os

sys.path.insert(0, os.path.abspath('..'))
from src.data_loader import cargar_datos
from src.preprocessing import (
    separar_caracteristicas_objetivo,
    dividir_conjunto,
    imputar_valores_faltantes,
    escalar_caracteristicas,
    codificar_variables_categoricas,
)

sns.set_theme(style='whitegrid')
```

## 1. Carga del conjunto de datos

```python
# Ejemplo con el dataset Iris
from sklearn.datasets import load_iris

iris = load_iris(as_frame=True)
df = iris.frame

# Para usar tu propio dataset, descomenta:
# df = cargar_datos('../data/raw/tu_archivo.csv')

print(f'Dimensiones: {df.shape}')
df.head()
```

## 2. Separación de características y variable objetivo

```python
columna_objetivo = 'target'  # Cambia al nombre de tu columna objetivo

X, y = separar_caracteristicas_objetivo(df, columna_objetivo)
print(f'Características: {X.shape} | Objetivo: {y.shape}')
print(f'Distribución del objetivo:\n{y.value_counts()}')
```

## 3. División en conjuntos de entrenamiento y prueba

```python
X_train, X_test, y_train, y_test = dividir_conjunto(
    X, y, test_size=0.2, random_state=42
)
print(f'Train: {X_train.shape} | Test: {X_test.shape}')
```

## 4. Imputación de valores faltantes

```python
X_train, X_test = imputar_valores_faltantes(
    X_train, X_test,
    estrategia_numerica='mean',
    estrategia_categorica='most_frequent',
)

print('Valores nulos en train después de imputar:', X_train.isnull().sum().sum())
print('Valores nulos en test después de imputar :', X_test.isnull().sum().sum())
```

## 5. Codificación de variables categóricas

```python
X_train, X_test = codificar_variables_categoricas(X_train, X_test)
print(f'Dimensiones tras codificación — Train: {X_train.shape} | Test: {X_test.shape}')
```

## 6. Escalado de características

```python
X_train_scaled, X_test_scaled, scaler = escalar_caracteristicas(X_train, X_test)

print('Estadísticas tras escalado (train):')
print(pd.DataFrame(X_train_scaled).describe().round(4))
```

## 7. Exportación de datos preprocesados

```python
os.makedirs('../data/processed', exist_ok=True)

X_train_scaled.to_csv('../data/processed/X_train.csv', index=False)
X_test_scaled.to_csv('../data/processed/X_test.csv', index=False)
y_train.to_csv('../data/processed/y_train.csv', index=False)
y_test.to_csv('../data/processed/y_test.csv', index=False)

print('Datos exportados en data/processed/')
```
