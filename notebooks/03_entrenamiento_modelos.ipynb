# 03 – Entrenamiento de Modelos

Este notebook entrena múltiples modelos de Machine Learning usando los datos preprocesados:
- Carga de datos preprocesados
- Entrenamiento de varios algoritmos
- Comparación rápida de desempeño en entrenamiento
- Guardado de los modelos entrenados

```python
import pandas as pd
import numpy as np
import sys
import os

sys.path.insert(0, os.path.abspath('..'))
from src.models import entrenar_modelos, guardar_modelo
```

## 1. Carga de datos preprocesados

> Ejecuta primero el notebook `02_preprocesamiento.ipynb` para generar estos archivos.

```python
X_train = pd.read_csv('../data/processed/X_train.csv')
X_test  = pd.read_csv('../data/processed/X_test.csv')
y_train = pd.read_csv('../data/processed/y_train.csv').squeeze()
y_test  = pd.read_csv('../data/processed/y_test.csv').squeeze()

print(f'Train: {X_train.shape} | Test: {X_test.shape}')
print(f'Clases: {sorted(y_train.unique())}')
```

## 2. Entrenamiento de modelos de clasificación

> Para regresión cambia `tarea='regresion'`.

```python
modelos_entrenados = entrenar_modelos(
    X_train, y_train, tarea='clasificacion'
)
```

## 3. Exactitud rápida en el conjunto de entrenamiento

```python
from sklearn.metrics import accuracy_score

for nombre, modelo in modelos_entrenados.items():
    acc_train = accuracy_score(y_train, modelo.predict(X_train))
    acc_test  = accuracy_score(y_test,  modelo.predict(X_test))
    print(f'{nombre:<30}  Train: {acc_train:.4f}  |  Test: {acc_test:.4f}')
```

## 4. Guardar modelos entrenados

```python
for nombre, modelo in modelos_entrenados.items():
    nombre_archivo = nombre.replace(' ', '_').lower()
    guardar_modelo(modelo, nombre_archivo, directorio='../models')
```
